```python
#1. remove values(rows) also called CCA --- dropna
#when to use CCA?
#1. if data MCAR(random missing) not like pattern - like last 20 missing, 1st 20 missing, rather should be missing randomly.
#2. if less than 5% missing. if more missing dont use CCA.

#3. also check after removing missing values the distribution. should remain same not alot change bcz we random removed.
#if before distribution and after distribution changes alot after removing the missing,thats a problem then.
#4. #for categorical variables check before after ratio. 

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Data/data_science_job.csv')
df.head()
```

```python
# each column percent missing
df.isnull().mean()*100

df.shape

#columns with less than 5% missing
cols = [var for var in df.columns if df[var].isnull().mean() < 0.05 and df[var].isnull().mean() > 0]
#Go through each column (var). If the condition is true, put var(that column) into the new list."
#gives in % bcz mean used.

#another way to do the same is
cols = []

for var in df.columns:
    if df[var].isnull().mean() < 0.05 and df[var].isnull().mean() > 0:
        cols.append(var)


cols
#it will return column names with less than 5% missing.

df[cols].sample(5) #created new df with only these columns

#lets see how much data will be left after removing missing from above columns..89% something
len(df[cols].dropna()) / len(df)     #total values in new data set "DIVIDE BY" total data in old set.

#after removing missing left new df.
new_df = df[cols].dropna()
df.shape, new_df.shape




#now plot individual columns before after distributions check -- hist plot so to prove that its missing at random if distri remains same.
#plot k uper plot banao

new_df.hist(bins=50, density=True, figsize=(12, 12))
plt.show()
```
